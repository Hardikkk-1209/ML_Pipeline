## 4. Careful words: what I can and cannot claim

### What I can claim

I can describe patterns and associations observed in the available data. For example, I can report that a particular signal is associated with higher or lower visibility, clicks, engagement, or performance movement within the analyzed data. I can also compare the strength of different signals and identify patterns that may be useful for further investigation.

### What I cannot claim

I cannot claim that an observed relationship proves that one signal causes another. I cannot claim that my analysis has discovered or proven a Google ranking algorithm factor. I also cannot claim that changing a particular signal will definitely improve a page's performance.

The results should therefore be described using careful language such as "observed," "associated with," "suggests," and "directional." The final purpose is decision support rather than replacing human judgment.
